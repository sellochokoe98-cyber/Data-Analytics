```python
columns = [
    "Diabetes mellitus (yes/no)",
    "Coronary artery disease (yes/no)",
    "Pedal edema (yes/no)",
    "Anemia (yes/no)"
]

for col in columns:
    
    print("\n", col)
    print(df[col].value_counts())

    df[col].value_counts().plot(kind="bar")

    plt.title(col)
    plt.xlabel("Response")
    plt.ylabel("Number of Patients")

    plt.show()
```

```python
for col in columns:

    df[col].value_counts().plot(
        kind="pie",
        autopct="%1.1f%%"
    )

    plt.title(col)
    plt.ylabel("")

    plt.show()
```
