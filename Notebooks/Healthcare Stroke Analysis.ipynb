```python
import pandas as pd
import matplotlib.pyplot as pltimport 
```

```python
df = pd.read_csv(r"C:\Users\sello\Downloads\healthcare-dataset-stroke-data.csv")

df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['bmi'] = df['bmi'].fillna(df['bmi'].mean())
```

```python
df.isnull().sum()
```

```python
df['stroke'] = df['stroke'].replace({
    0: 'No Stroke',
    1: 'Stroke'
})
```

```python
df['hypertension'] = df['hypertension'].replace({
    0: 'No',
    1: 'Yes'
})
```

```python
df['heart_disease'] = df['heart_disease'].replace({
    0: 'No',
    1: 'Yes'
})
```

```python
df.head()
```

```python
df['stroke'].value_counts().plot(kind='bar')

plt.title('Stroke Cases')
plt.xlabel('Stroke Status')
plt.ylabel('Number of People')

plt.show()
```

```python
df['hypertension'].value_counts().plot(kind='bar')

plt.title('Hypertension Status')
plt.xlabel('Hypertension')
plt.ylabel('Count')

plt.show()
```

```python
df['heart_disease'].value_counts().plot(kind='bar')

plt.title('Heart Disease Status')
plt.xlabel('Heart Disease')
plt.ylabel('Count')

plt.show()
```

```python
pd.crosstab(df['gender'], df['stroke']).plot(kind='bar')

plt.title('Stroke by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.show()
```

```python
pd.crosstab(df['smoking_status'], df['stroke']).plot(kind='bar')

plt.title('Stroke by Smoking Status')
plt.xlabel('Smoking Status')
plt.ylabel('Count')

plt.xticks(rotation=45)

plt.show()
```

```python
df.groupby('stroke')['avg_glucose_level'].mean().plot(kind='bar')

plt.title('Average Glucose Level')
plt.xlabel('Stroke Status')
plt.ylabel('Average Glucose')

plt.show()
```

```python
df['work_type'].value_counts().plot(kind='bar')

plt.title('Work Type Distribution')
plt.xlabel('Work Type')
plt.ylabel('Count')

plt.xticks(rotation=45)

plt.show()
```
